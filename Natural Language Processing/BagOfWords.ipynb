```python
import pandas as pd

# Try reading the file with a different encoding, like 'latin1'
data = pd.read_csv('spam.csv', encoding='latin1')

selected_data = data[['v1','v2']]#Selecting the particular columns only
```

### Data cleaning and Pre-Processing

```python
import nltk
import re
nltk.download('stopwords')
```

```python
nltk.download('wordnet')
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
corpus = []
for i in range (0,len(selected_data)):
  review = re.sub('[^a-zA-Z]',' ',selected_data['v2'][i])#Apart a to z both small and large replace with empyty space
  review.lower()
  review = review.split()
  review = [lemmatizer.lemmatize(words) for words in review if words not in set(stopwords.words('english'))]
  review = ' '.join(review)
  corpus.append(review)
  print(review)
```

## Creating Bag Of Words

```python
import sklearn   #https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features = 500, binary=True, ngram_range=(1,2))
```

```python
Vectors = cv.fit_transform(corpus).toarray()
```
